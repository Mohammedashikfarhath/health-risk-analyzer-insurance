# Health Risk Analyzer — Insurance Charges EDA

Analyze how **age, BMI, smoking, children, region** drive insurance `charges`.
Dataset: `data/insurance.csv` (1338 rows). Run top-to-bottom in Jupyter.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme(style='whitegrid')

df = pd.read_csv('../data/insurance.csv')
print(df.shape)
display(df.head())
display(df.describe())
print(df.isnull().sum())
```

## 1. Cleaning + Feature Engineering

```python
import sys; sys.path.append('../src')
from data_cleaning import clean_data, summary_tables

cdf = clean_data(df)
cdf.to_csv('../data/insurance_clean.csv', index=False)
print(cdf.shape)
display(cdf.head())
```

## 2. Univariate + Bivariate

```python
# Charges distribution
sns.histplot(cdf['charges'], kde=True, bins=30)
plt.title('Distribution of Charges'); plt.show()

# Smoker vs charges
sns.boxplot(data=cdf, x='smoker', y='charges')
plt.title('Smokers vs Non-smokers'); plt.show()

# Age & BMI vs charges
sns.scatterplot(data=cdf, x='age', y='charges', hue='smoker', alpha=0.7)
plt.title('Age vs Charges'); plt.show()
sns.scatterplot(data=cdf, x='bmi', y='charges', hue='smoker', alpha=0.7)
plt.title('BMI vs Charges'); plt.show()
```

## 3. Aggregations

```python
tables = summary_tables(cdf)
for k, t in tables.items():
    print(f'\n== {k} ==')
    display(t)

print(cdf[['age','bmi','children','charges','is_smoker','risk_score']].corr()['charges'])
```

## Key Takeaways
- Smokers ~3.8x higher mean charges (~$32k vs ~$8.4k), corr 0.79 — dominant driver.
- Risk score (smoker+obese+senior) gradient: 0=$6.2k → 3=$47k.
- Age corr 0.30, BMI corr 0.20 — secondary drivers; Obese avg $15.5k.
- Region minimal effect; Southeast slightly highest.
